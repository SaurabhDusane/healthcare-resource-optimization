# 02 — Modeling & Forecasting

Runs the end-to-end pipeline on synthetic data and inspects the acuity
classifier and the demand forecast. All metrics are illustrative (synthetic
data), reproducible from the seed.

> Install the lightweight stack first: `pip install -r ../requirements-ci.txt`

```python
import os, sys, json
sys.path.insert(0, os.path.abspath('..'))
from src.pipeline import Pipeline, PipelineConfig

metrics = Pipeline(PipelineConfig(n_visits=6000, model_type='random_forest')).run()
print('data source:', metrics['data_source'])
```

## Acuity classifier

```python
acuity = metrics['acuity_model']
print('metrics :', acuity['metrics'])
print('registry:', acuity['registry_version'])
for f in acuity['top_features']:
    print(f"  {f['feature']:20s} {f['importance']:.3f}")
```

## Demand forecast (backtest, CV, exogenous signals)

```python
fc = metrics['forecast']
print('best model   :', fc['best_model'], fc['best_metrics'])
print('cv mean MAE  :', fc['cross_validation']['cv_mean_mae'])
print('exog helps   :', fc.get('exogenous', {}).get('exog_helps'))
```
